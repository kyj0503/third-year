```python
word = input('단어를 입력하세요: ')

reverse_word = ''
for i in word:  # 입력받은 단어의 각 문자를 순회한다.
    reverse_word = i + reverse_word

if word == reverse_word:
    print('회문입니다.')
else:
    print('회문이 아닙니다.')
```

```python
word = input('단어를 입력하세요: ')

left = 0  # 시작 인덱스 초기화
right = len(word) - 1  # 마지막 인덱스 초기화

is_palindrome = True  # 회문 여부를 저장할 변수 초기화
while left < right:  # 왼쪽 인덱스가 오른쪽 인덱스보다 작을 때까지 반복
    # 양 끝의 문자가 같은지 비교합니다.
    if word[left] == word[right]:  # 같다면 left를 1 증가시키고 right를 1 감소시킨다.
        left += 1
        right -= 1
        continue
    else:  # 다르다면 is_palindrome을 False로 변경하고 반복문을 빠져나온다.
        is_palindrome = False
        break

print(is_palindrome)
```

```python
t = 'hello world' # text
p = 'wor'         # target pattern

def find_word(p, t):
    N = len(t)
    M = len(p)

    # 텍스트 t에서 패턴 p를 검색할 수 있는 모든 시작 위치를 순회한다.
    # N-M+1은 패턴이 텍스트를 벗어나지 않도록 하기 위한 범위이다.
    for i in range(N-M+1):  
        cnt = 0  # 현재 위치에서 일치하는 문자 수를 세기 위한 변수
        for j in range(M):  # 패턴 p의 각 문자를 순회한다.
            if t[i+j] == p[j]:  # 텍스트 t의 현재 위치에서 패턴 p의 문자와 일치하는지 확인한다.
                cnt += 1  # 일치하면 cnt 증가
            else:
                break  # 일치하지 않으면 루프 종료

        if cnt == M:  # cnt가 패턴의 길이와 같으면 패턴이 완전히 일치한 것이므로 현재 시작 위치 i를 반환
            return i

    return '못찾았음'

print(find_word(p, t))
```
