집계

```python
gems = [3,3,1,2,3,2,2,3,3,1]
grades = {1:0, 2:0, 3:0}

for i in gems:
    grades[i] += 1

print(grades)
```

```python
gems = [3,3,1,2,3,2,2,3,3,1]
result = sum(gems)

if result <= 15:
    print('성공')
elif 15 < result <= 23:
    print('보통')
else:
    print('실패')
```

선형 탐색 루 최솟값, 최댓값 구하기

```python
nums = [7,1,2,4,6,8,3]

#최댓값
max_num = -1
for num in nums:
    if max_num < num:
        max_num = num

print(max_num)

#최솟값
min_num = 9999
for num in nums:
    if min_num > num:
        min_num = num
print(min_num)
```
