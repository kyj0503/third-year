피보나치

```python
memo = [0,1]

def fibo(n):
    if n >= 2 and n >= len(memo):
        memo.append(fibo(n-1) + fibo(n-2))
    return memo[n]

print(fibo(10))
```

*이진검색
반드시 정렬된 배열에서만 사용 가능
배열의 중간 값을 가져오고, 중간 값과 검색 값을 비교한다.
중간 값이 검색 값과 같으면 종료.
중간 값보다 검색 값이 크다면 중간 값 기준 배열의 오른쪽 구간을 대상으로 탐색한다.
중간 값보다 검색 값이 작다면 중간 값 기준 배열의 왼쪽 구간을 대상으로 탐색한다.
값을 찾거나 간격이 비어있을 때까지 반복한다.

```python
nums = [1,2,3,4,5,6,7,8,9]

def binary_search(low, high, target):
    if low > high:
        return '찾지 못함'

    mid = (low + high) // 2
    if target == nums[mid]:
        return mid
    elif target < nums[mid]:
        return binary_search(low, mid-1, target)
    elif target > nums[mid]:
        return binary_search(mid+1, high, target)

print(binary_search(0, len(nums)-1, 7))
```

시작 값, 끝 값, 목표 값 넣으면 목표 값의 인덱스 반환

실절 백트래킹 연습문제
https://swexpertacademy.com/main/code/problem/problemDetail.do?contestProbId=AV5PpFQaAQMDFAUq

```python
T = int(input()) # 테스트 케이스의 수

def find_cheapest(month, acc): # month는 현재 탐색 중인 달. acc는 현재 누적비용
    global answer # 가장 저렴한 비용을 저장하는 변

    if acc >= answer: # 중간 백트래킹 : 이미 너무 비싼 루트면 가지치기
        return

    if month > 11: # 범위를 벗어난 경우. 12월까지 모두 구매해본 경우
        if answer > acc:
            answer = acc # 현재 값으로 누적 비용 갱신하기
        return

    # 1일권 plan 수만큼 사서 다음달로
    find_cheapest(month+1, acc + plans[month]*prices[0]) # prices[0]은 1일권 가격 * 일수

    # 1달권 사서 다음달로
    find_cheapest(month+1, acc + prices[1]) # prices[1]은 1달권 가격

    # 3달권 사서 다음달로
    find_cheapest(month+3, acc + prices[2]) # prices[2]은 3달권 가격

for tc in range(1, T+1):
    prices = list(map(int, input().split()))
    plans = list(map(int, input().split()))
    answer = prices[3] # 일단 1년권 가격을 기준으로 잡음
    find_cheapest(0, 0)

    print('#{} {}'.format(tc, answer))
```

map(function, iterable) 함수는 iterable의 각 요소에 대해 function 함수를 적용한 결과를 새로운 iterator로 반환한다.
이때 function 함수는 각 요소를 인자로 받아서 처리하며, 함수의 반환값이 새로운 iterator의 각 요소가 됩니다.

iterable 이터러블 - 반복 가능한 데이터, 객체. 순서형 및 컬렉션 자료형.
string, list, tuple, range(), set, dict
